```python
import numpy as np
import matplotlib.pyplot as plt

# Rentang x1
x = np.linspace(0, 2000, 1000)

# Kendala
y1 = (30000 - 12*x)/24      # 12x1 + 24x2 <= 30000
y2 = (27000 - 18*x)/12      # 18x1 + 12x2 <= 27000
y3 = (1350 - 0.75*x)/0.8    # 0.75x1 + 0.8x2 <= 1350

# Hanya tampilkan nilai positif
y1 = np.maximum(y1, 0)
y2 = np.maximum(y2, 0)
y3 = np.maximum(y3, 0)

plt.figure(figsize=(8,6))

# Garis kendala
plt.plot(x, y1, label=r'$12x_1+24x_2\leq30000$')
plt.plot(x, y2, label=r'$18x_1+12x_2\leq27000$')
plt.plot(x, y3, label=r'$0.75x_1+0.8x_2\leq1350$')

# Titik sudut daerah feasible
feasible_x = [0, 0, 750, 1000]
feasible_y = [0, 1125, 750, 0]

# Arsiran daerah feasible
plt.fill(feasible_x, feasible_y, color='gray', alpha=0.3,
         label='Daerah Feasible')

# Titik optimal
plt.scatter(750, 750, s=80)
plt.annotate('(750,750)', (750,750),
             textcoords="offset points",
             xytext=(10,10))

plt.xlim(0, 2000)
plt.ylim(0, 2000)

plt.xlabel(r'$x_1$')
plt.ylabel(r'$x_2$')
plt.title('Daerah Feasible dan Solusi Optimal')
plt.grid(True)
plt.legend()

plt.savefig('grafik_feasible.png', dpi=300,
            bbox_inches='tight')

plt.show()
```
